```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

# emparejamos cada variable promedio con su respectiva mediana
variables = list(zip(variables_promedio, variables_mediana))
n_vars = len(variables)

# definimos cantidad de columnas y filas dinámicamente
n_cols = 3  # puedes ajustarlo si prefieres 2 o 4 columnas
n_rows = math.ceil(n_vars / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(6*n_cols, 5*n_rows), sharey=False)
axes = axes.flatten()

for ax, (var_prom, var_med) in zip(axes, variables):
    # dataframe combinado de promedio y mediana
    df_plot = df_resumen[['grupo', var_prom, var_med]].melt(
        id_vars='grupo',
        value_vars=[var_prom, var_med],
        var_name='Estadístico',
        value_name='valor'
    )
    df_plot['Estadístico'] = df_plot['Estadístico'].map({
        var_prom: 'Promedio',
        var_med: 'Mediana'
    })

    # boxplots
    sns.boxplot(
        data=df_plot, x='grupo', y='valor', hue='Estadístico',
        palette={'Promedio': 'lightskyblue', 'Mediana': 'lightseagreen'},
        ax=ax
    )

    ax.set_title(var_prom.replace("_promedio",""), fontsize=14)
    ax.set_xlabel('Grupo')
    ax.set_ylabel('Valor')
    ax.tick_params(axis='x', rotation=45)

# eliminar ejes vacíos si sobran
for ax in axes[n_vars:]:
    ax.set_visible(False)

plt.tight_layout()
#guardar_figura('promedioymedianaporvariable', 'Promedio y mediana')
plt.show()
```

## Visualización por salinidad para cada variable (continua, por grupo)

```python
df['salinidad'] = df['salinidad'].astype(str)
```

```python
alta_palette = [ '#ffb3b3', '#ff9999', '#ff6666','#e64d4d', '#cc3333']
normal_palette = [  "#9a9aff", "#6464fe", "#3030c9"]
baja_palette = [  "#99ff99ff", "#56ef56e7", "#50d750", "#24b124" ]


# Paletas suaves y distinguibles para cada nivel
paletas_por_nivel = {
    'baja': baja_palette,
    'normal': normal_palette,
    'alta': alta_palette
}


# Extraer salinidad y nivel únicos
niveles_salinidad = df[['salinidad', 'nivel_salinidad']].drop_duplicates()

# Mapa: salinidad → color
mapa_colores = {}

for nivel in ['baja', 'normal', 'alta']:
    subniveles = niveles_salinidad[niveles_salinidad['nivel_salinidad'] == nivel]['salinidad'].tolist()
    paleta = paletas_por_nivel[nivel]
    if len(subniveles) != len(paleta):
        print(f"[Advertencia] En el nivel '{nivel}', hay {len(subniveles)} subniveles y {len(paleta)} colores. Verifica que coincidan.")

    for s_val, color in zip(subniveles, paleta):
        mapa_colores[s_val] = color  # ya es str
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def graficar_kde_por_salinidad_y_variable(df, variables, orden_niveles_salinidad, nombre, hue=None):
    n_filas = len(variables)
    n_columnas = len(orden_niveles_salinidad)

    fig, axes = plt.subplots(nrows=n_filas, ncols=n_columnas,
                             figsize=(5 * n_columnas, 3 * n_filas),
                             sharex=False, sharey=False)
    axes = np.atleast_2d(axes)

    for i, variable in enumerate(variables):
        if variable not in df.columns:
            print(f"[Advertencia] Variable '{variable}' no está en el DataFrame. Se omite.")
            continue

        max_y_variable = 0
        x_vals_min = np.inf
        x_vals_max = -np.inf

        # dibujar curvas y calcular límites
        for j, s in enumerate(orden_niveles_salinidad):
            subset = df[df['nivel_salinidad'] == s]
            ax = axes[i, j]

            if hue is not None:
                # niveles de hue presentes en este subplot, convertidos a str
                niveles_hue = sorted(subset[hue].astype(str).unique())
                #colores = sns.color_palette(palette, len(niveles_hue))
                colores = [mapa_colores[str(h_val)] for h_val in niveles_hue]

                for k, h_val in enumerate(niveles_hue):
                    sub_h = subset[subset[hue].astype(str) == h_val]
                    sns.kdeplot(data=sub_h, x=variable, ax=ax, color=colores[k], common_norm=False, label=str(h_val))
            else:
                sns.kdeplot(data=subset, x=variable, ax=ax)

            # límites Y
            max_y_subplot = max(line.get_ydata().max() for line in ax.lines)
            if max_y_subplot > max_y_variable:
                max_y_variable = max_y_subplot

            # límites X basados en curvas
            for line in ax.lines:
                x_line = line.get_xdata()
                x_vals_min = min(x_vals_min, x_line.min())
                x_vals_max = max(x_vals_max, x_line.max())

            # agregar leyenda de este subplot
            if hue is not None:
                ax.legend(title=hue, loc='upper right', fontsize=8)

        # ajustar todos los subplots de la variable
        for j, s in enumerate(orden_niveles_salinidad):
            ax = axes[i, j]
            ax.set_ylim(0, max_y_variable * 1.1)
            ax.set_xlim(x_vals_min, x_vals_max)
            ax.set_title(f"{variable} - Salinidad {s}")
            ax.set_xlabel('')
            ax.set_ylabel('')

        # etiquetas generales por fila y columna
        axes[i, 0].set_ylabel("KDE")
        for ax in axes[i, :]:
            ax.set_xlabel("Valor")

    plt.tight_layout()
    guardar_figura(nombre, 'kde_por_variable_V3')
    plt.show()
```

## Correlaciones por salinidad

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

def graficar_heatmaps_correlacion_por_salinidad(df, variables, orden_niveles_salinidad, nombre):
    n_columnas = len(orden_niveles_salinidad)
    fig, axes = plt.subplots(nrows=1, ncols=n_columnas, figsize=(6 * n_columnas, 6), squeeze=False)

    vmin, vmax = -1, 1  # misma escala para todos
    cbar_ax = fig.add_axes([0.92, 0.155, 0.02, 0.785])  # posición del colorbar único

    for i, s in enumerate(orden_niveles_salinidad):
        subset = df[df['nivel_salinidad'] == s]
        corr = subset[variables].corr(method='spearman')

        sns.heatmap(
            corr,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            center=0,
            vmin=vmin,
            vmax=vmax,
            linewidths=0.5,
            cbar=(i == n_columnas - 1),      # solo el último con colorbar
            cbar_ax=None if i != n_columnas - 1 else cbar_ax,
            ax=axes[0, i]
        )
        axes[0, i].set_title(f"Salinidad {s}")

        # mostrar nombres de variables solo en el primer heatmap
        if i == 0:
            axes[0, i].set_yticklabels(variables, rotation=0)
        else:
            axes[0, i].set_yticklabels([])
            axes[0, i].set_ylabel('')

        # quitar ticks de todos los heatmaps
        axes[0, i].tick_params(left=False, labelleft=(i==0))

    cbar_ax.set_ylabel('Correlación de Spearman', rotation=270, labelpad=15)
    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.subplots_adjust(wspace=0.15) # deja espacio para el colorbar
    guardar_figura(nombre, 'heatmaps_correlacion_por_salinidad_V3')
    plt.show()
```
