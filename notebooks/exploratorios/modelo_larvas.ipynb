Para $r$ es un modelo CIR; para $\theta$ un Orsein-Uhlenbeck.

\begin{equation}
    \begin{aligned}
        x_t &= r_t\cos(\theta_t)\\
        y_t &= r_t\sin(\theta_t)\\
        r_t &= r_0 + \int_0^t\alpha_r(r_\infty-r_s)ds + \int_0^t \sigma_r\sqrt{r_s}dB^{(r)}_s\\
        \theta_t &= \theta_0+ \int_0^t\alpha_\theta(\theta_\infty-\theta_s)ds + \int_0^t \sigma_\theta dB^{(\theta)}_s
    \end{aligned}
\end{equation}

```python
# ID 45

M = 10
T = 11.51
n = 5000
dt = T/n
t = np.arange(0,T+dt,dt)
nt = len(t)

r0 = 0.018990733108545495
theta0= -0.6407102038616406
theta_dot0= -0.27062238283239415

alpha_r = 1.1500000000000006e-10
r_infty = 9.749999999999996e-11
sigma_r = 0.06372633423360888
alpha_theta = 21.408729868420846
theta_dot_infty = 0.38985927842287205
sigma_theta = 11.458414444942497

theta = np.zeros(shape=(nt,M))
theta_dot = np.zeros(shape=(nt,M))
r = np.zeros(shape=(nt,M))

theta[0,:] = theta0 * np.ones(M)        # todos parten del ángulo observado
theta_dot[0,:] = theta_dot0 * np.ones(M)  # velocidad angular inicial
r[0,:] = r0 * np.ones(M)                  # radio inicial observado

# theta[0,:] = np.random.uniform(0,2*np.pi,size=M) #theta0*np.ones_like(theta[0,:])
# theta_dot[0,:] = theta_dot0*np.zeros_like(theta_dot[0,:])
# r[0,:] = r0*np.random.uniform(0,1,size=M)

for k in range(nt-1):
    theta[k+1,:] = theta[k,:] + dt * theta_dot[k,:] 
    theta_dot[k+1,:] = theta_dot[k,:] + alpha_theta * (theta_dot_infty-theta_dot[k,:] )*dt + sigma_theta * np.sqrt(dt) * np.random.standard_normal(size=(1,M))
    r[k+1,:] = r[k,:] + alpha_r * (r_infty - r[k,:]) * dt + sigma_r * np.sqrt(dt) * r[k,:] * np.random.standard_normal(size=(1,M))# np.sqrt(np.abs(r[k,:])) * np.random.standard_normal(size=(1,M))
    
x = r * np.cos(theta)
y = r * np.sin(theta)

plt.plot(x,y);
```
